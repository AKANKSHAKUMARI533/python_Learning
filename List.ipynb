**List Collection**
A list is an ordered, mutable (changeable) collection used to store multiple values.
- Allow duplicates
- Can store mixed data types

**List are used used to:**
* Store datasets
* Hold features & labels
* Preprocess raw data
* Feed data to NumPy/Pandas
* Batch processing in ML




```python
#List
nums = [10, 20, 30]
data = [1,'AI', 3.5, True]
empty = []
print(nums)
print(data)
print(type(nums),type(data),type(empty))
```

**List constructor** - used to create a new list object, often by taking an exisitng iterables (like a tuple, string, or set)

```python
#Convert iterables into lists.
my_tuple = (1,2,3)
my_list = list(my_tuple)
print(my_tuple,type(my_tuple),'\n',my_list,type(my_list),'\n')

#Create empty list
empty_list = list()
print(empty_list)

#Copying
copy_list = list(my_list)
print(copy_list)
```

**Indexing** - Accessing List elements

```python
nums = [10,20,30,40,50]
f1 = nums[0]    #10
f2 = nums[-1]   #50
f3 = nums[1]    #20
print(f1,f2,f3)
```

**List Slicing** - It uses a special [start:stop:step] syntax to create a new sublist containing only the elements within the specified range.

```python
print(nums[1:3])  #From 1st index to 3rd index, 3rd not included
print(nums[:3])   #From 0 to till 3rd index, 3rd is not included
print(nums[2:])   #start from 2 index till end
print(nums[::-1]) #reverse
```

**Updating List Elements**

```python
nums[1] = 199
print('updated list', nums)

#Append - Add one element
nums.append(1999)
print('Appended list', nums)

#Extend - Add multiple element
nums.extend([140,120])
print('extended list', nums)

#insert(index,value) - Add at index
nums.insert(4,70)
print('inserted list', nums)
```

**Removing Element**

```python
print('nums before removing, poping and pop by index ', nums)
nums.remove(1999)   #Remove by value
nums.pop()          #remove last
nums.pop(1)         #remove by index
print('nums after removing, poping and pop by index ', nums)
nums.clear()
print('nums after clear ', nums)
```

**Serching & Counting**


```python
nums = [10,20,30,40,50,60,30,40]
print(nums)

print(nums.index(50)) # By value serching the index place
print(nums.count(30)) # Count a paticular value in list
```

**Sorting & Reversing**

```python
nums.sort() #Sort a list in order
print(nums)

nums.sort(reverse=True) #Sort a list in descending order
print(nums)

NList = [5,3,4,6,2,7]
NList.reverse()  #Reverse the list, no sorting
print(NList)

print(sorted(NList), NList) #safe sorting, Not modifying the original list
```

**List Operators**


```python
a = [12, 15, 18]
b = [2,5,6]
print(a + b)  #concatenation
print(a * 2)  #repetion
```

**Membership & Length**

```python
nums = [10,20,30,40,50]
print(1 in nums)
print(20 in nums)
len(nums)
```

**Looping Through Lists**

```python
print('list loop')

for x in nums:
  print(x)

print('\n')
print('list loop with index')

#enumerate means mention (a number of things) one by one.
for i, val in enumerate(nums):
  print(i,val)
```

**List Comprehension**

```python
#built-in Python function that generates a sequence of numbers
#starting from 0 (by default) up to,
#but not including, the number provided as an argument.
listRange = range(5)
print(type(listRange))
print(listRange)

squares = [x*x for x in range(6)]
print(squares)

nums = [1,2,3,4,5,6,7,8]
even = [x for x in nums if x%2 == 0]
print(even)
```

**Nested List** - Matrices /Datasets

```python
matrix = [[1,2], [3,4]]
print(matrix)

print(matrix[1][1])
print(matrix[1][0])
```

**Copying List**

```python
a = [[1,2],[3,4]]

#Assignemnt Reference - This does not  create a new list
#it makes ab a new reference to the same memory object that a points to
ab = a
ab.append(30) #if we modify ab, a will be also change
print(a,ab)

#Created independent list, affects modification No (Yes for nested items)
bc = a.copy()
bc.append(60)
print(bc, a)

#Created independent list, affects modification No (Yes for nested items)
cd = a[:]
cd.append(80)
print(cd,a)
```
