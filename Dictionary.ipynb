**Dictionary**
A dictionary is a collection of key-value pairs
* Key-value structure
* Mutable
* Keys must be unique
* Ordered (python 3.7+)
* No index access

Dictionaries are used for:
* Feature-label mapping
* JSON data handling
* API responses
* Model configurations
* Word frequency counting
* Dataset representation

👉 Almost every ML project uses dictionaries.

```python
d = {"a": 1, "b": 2}
empty = {}
dict_constructor = dict(name = 'AI', level =1)

print(d)
print(empty)
print(dict_constructor)
```

```python
student = {"name": "Akanksha","age": 26}

print(student["name"])
print(student.get("age"))

#Adding & Updating
student["city"] = "Delhi"
student["grade"] = "A+"

print(student)
print(type(student))
```

**Removing Items**

```python
student.pop("age")
print(student)

student.popitem()
print(student)

del student["city"]
print(student)

student.clear()
print(student)
```

**Dictionary Methods**

```python
Me = {'name': 'Akanksha', 'city': 'Delhi', 'grade': 'A+'}
age = {"age": 26}

print(Me.keys())    #Fetch all keys
print(Me.values())  #Fetch all values
print(Me.items())   #Get key-value pairs

print(Me.get("name"))   #Safe access

Me.update(age)    # Merge two dict
print(Me)
```

**Looping Through Dictionary**

```python
student = {'name': 'Akanksha', 'city': 'Delhi', 'grade': 'A+', 'age': 26}

for key in student:
  print(key)

print('\n')

for value in student.values():
  print(value)

print('\n')

for key, value in student.items():
  print(key, value)
```

**Dictionary in AI - Word Frequency**

👉 Used in NLP preprocessing.

```python
text = "AI is powerful and AI is future"

words = text.split()

freq = {}

for word in words:
  freq[word] = freq.get(word,0) + 1

print(freq)
```

**Dictionary Comprehension**
a concise and efficient way in Python to create a new dictionary or transform an existing one

```python
Squares = {x: x*x for x in range(5)}
print(Squares)

even_squares = {x: x*x for x in range(10) if x % 2 == 0}
print(even_squares)

d1 = {"a" : 1}
d2 = {"b" : 2}

merged = {**d1, **d2}
print(merged)

```
