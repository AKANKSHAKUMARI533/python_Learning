**Set Collection**
An unordered, Mutable collection of UNIQUE elements

**Key Properties**
* No duplicate values
* No indexing
* Unordered
* Mutable
* Fast memebership checking

**Sets are used for:**
* Removing duplicate data
* Feature comparison
* Finding common elements between datasets
* Fast lookup operations
* Text preprocessing (unique words)
👉 Very useful in data cleaning & NLP

```python
s = {1,2,3}
empty = set()   # {} creates a dictionary, not a set
```

**Duplicate Removal**

```python
nums = [1,2,2,4,4,6,7,6,7]
unique = set(nums)
print(unique)
```

**Adding Elements**

```python
s.add(4)  #Add single element
print(s)

s.update([5,6,7]) #Add muliple elements
print(s)
```

**Removing Elements**

```python
s.remove(2)  #error if not present
print(s)

s.discard(3) #No error if not present
print(s)

s.pop()      #Remove random element
print(s)

s.clear()    #Empty Set
print(s)
```

**Set Operations**

```python
A = {1,2,3}
B = {4,5,6}

#Union (All elements)
print(A|B)
print(A.union(B))

#Interaction (Common elements)
print(A&B)
print(A.intersection(B))

#Difference (Elements in A not in B)
print(A-B)

#Symmetric Difference
print(A^B)

#Memebership (Very Fast) important for large datasets
print(3 in A)

#Looping - order is not guaranteed
for item in A:
  print(item)
```

**Frozen Set** - is immutable

Used when:

**Sets in NLP** - Unique words in text
* Removes duplicate words
* Used in vocabulary building

```python
text = "AI is powerful and AI is future"
text1 = "AI is powerful and ai is future"

print(set(text.split()))
print(set(text1.split()))
```
