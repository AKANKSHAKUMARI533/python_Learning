**Tuple**
A tuple is an ordered, immutable (unchangeable) collection

* Ordered
* Allow duplicates
* Cannot modify after creation
* Faster than lists
* Can be used as dictionary keys

Tuples are used for:
* Returning multiple values from functions
* Representing fixed configurations
* Storing coordinates (x, y)
* Model outputs (loss, accuracy)
* Safe data that should not change

👉 In ML, many functions return tuples

👉 Use tuple when data should NOT change

Performance Insight (AI Level)
* Faster than lists
* More memory efficient
* Safer for model configs

Used in:
* NumPy shapes
* Pandas index structures
* ML parameter returns

```python
t = (1,2,3,4)
single = (5,)   #single element tuple
empty = ()

#Accessing Tuple Elements
t[0]
t[-1]
```

**Tuple Slicing**
Same slicing rules as lists.

```python
print(t[0:2])
print(t[:2])
print(t[::-1])
```

**Tuple Operations**

```python
t1 = (1,2)
t2 = (3,4)

#Concatenation
print(t1 + t2)

#Repetition
print(t1 * 2)

#Membership
print(3 in t1)
print(2 in t1)

#Length
len(t1)
```

**Tuple Methods**

```python
TM = (1,1,1,2,2,2,2,4,3,3,3,4)
print(TM)

print(TM.count(4))

print(TM.index(4))
```

**Tuple Packing & UnPacking**
used heavily in AI

```python
data = 1,2,3

print(data)
print(type(data))

a, b, c = data

print(data)
print(type(data))

print(a)
print(type(a))

print(b)
print(type(b))

print(c)
print(type(c))

loss, accuracy = (0.25, 92)

print(loss)
print(type(loss))

print(accuracy)
print(type(accuracy))
```

**Nested Tuples**
Used In:
* Image coordinates
* Model Layer Configurations

```python
matrix = ((1,2), (3,4))
matrix[1][0]
```

**Tuple As Dictionary Key**
Because tuples are immutable
Lists cannot be used as keys.
Used in:
* Coordinates mapping
* Feature pairing

```python
d = {(1,2): "point"}

print(d)

#Converting between list & tuple
t = tuple([1,2,3,4])
print(t,type(t))

l = list((1,2,3,4,5,6))
print(l,type(l))
```
